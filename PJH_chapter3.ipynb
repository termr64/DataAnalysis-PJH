**[PJH 3-22] 데이터 확인**

```python
import matplotlib.pyplot as plt
import pandas as pd
df=pd.read_csv('temp_PJH.csv',encoding='cp949')
df.head(3)
```

**[PJH 3-23] 데이터프레임 개요 확인**

```python
df.info()
```

**[PJH 3-24] 결측치 채우기**

```python
df2= df.fillna(method='ffill')
df2.info()
```

**[PJH 3-25] 특수 문자가 포함된 열 이름을 변경**

```python
df2.rename(columns={'평균기온(°C)':'avg_temp'},inplace=True)
df2.rename(columns={'최고기온(°C)':'max_temp'},inplace=True)
df2.info()
```

**[PJH 3-26] 일시 데이터를 날짜형으로 변경**

```python
df2['일시'] = pd.to_datetime(df2['일시'],format='%Y-%m-%d')
df2.info()
```

**[PJH 3-27] 일시 데이터를 날짜형으로 변경**

```python
df3=df2.set_index('일시')
df3.head(3)
```

**[PJH 3-28] 9월 데이터 추출**

```python
df_sep = df3[pd.DatetimeIndex(df3.index).month == 9]
df_sep
```

**[PJH 3-29] 9월 기온 꺾은선 그래프 생성**

```python
plt.title('Seoul 2024, 9')
plt.plot(range(0,30),df_sep['max_temp'],label='max',c='r')
plt.plot(range(0,30),df_sep['avg_temp'],label='avg',c='y')
plt.xlim(1,30)
plt.ylim(20,37)
plt.legend()
plt.rcParams['figure.figsize']=(8,8)
plt.show()
```
